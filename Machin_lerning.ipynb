```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier

num_cols = ['Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
cat_cols = ['Sex', 'Embarked']

num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median'))
])

cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('ohe', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer([
    ('num', num_pipeline, num_cols),
    ('cat', cat_pipeline, cat_cols)
])

pipe = Pipeline([
    ('preprocessor', preprocessor),
    ('model', DecisionTreeClassifier(random_state=42))
])

pipe.fit(X_train, y_train)

y_pred = pipe.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

print(accuracy_score(y_test, y_pred))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import FunctionTransformer

df = pd.read_csv('/content/titanic.csv')

# Handle missing values
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Fare'] = df['Fare'].fillna(df['Fare'].median())

df[['Age', 'Fare']].head()
```

```python
log_transformer = FunctionTransformer(np.log1p)

df['Fare_log'] = log_transformer.fit_transform(df[['Fare']])

df[['Fare', 'Fare_log']].head()
```

```python
#Square Root Transformation
#Concept: Reduce the impact of large values.


sqrt_transformer = FunctionTransformer(np.sqrt)

df['Fare_sqrt'] = sqrt_transformer.fit_transform(df[['Fare']])

df[['Fare', 'Fare_sqrt']].head()
```

```python
#Reciprocal Transformation
#Concept: Apply the reciprocal of each value.

reciprocal_transformer = FunctionTransformer(
    lambda x: 1 / x
)

# Fare must be greater than zero
df['Fare_reciprocal'] = reciprocal_transformer.transform(
    df[['Fare']].replace(0, np.nan)
)

df[['Fare', 'Fare_reciprocal']].head()
```

```python
import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer

df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Male',
               np.nan, 'Male', 'Female', np.nan]
})

imputer = SimpleImputer(strategy='most_frequent')

df['Gender'] = imputer.fit_transform(df[['Gender']]).ravel()

print(df)
```

```python
import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer

df = pd.DataFrame({
    'City': ['Kolkata', 'Delhi', 'Mumbai',
             np.nan, 'Kolkata', np.nan]
})

imputer = SimpleImputer(
    strategy='constant',
    fill_value='Missing'
)

df['City'] = imputer.fit_transform(df[['City']]).ravel()

print(df)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Age': [20, 25, 30, np.nan, 35, np.nan, 40]
})

# Store non-missing values
random_sample = df['Age'].dropna().sample(
    df['Age'].isnull().sum(),
    random_state=42
)

# Replace missing values
df.loc[df['Age'].isnull(), 'Age'] = random_sample.values

print(df)
```

```python
#Missing indicato

# Missing Indicator is a technique used in Machine Learning to create a new binary column that indicates whether the original value was missing or not.

# It does not fill missing values. Instead, it records where missing values existed.
# - 1 → Value was missing.
# - 0 → Value was present.
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Age': [20, 25, np.nan, 30, np.nan, 40]
})

# Create missing indicator
df['Age_NA'] = df['Age'].isnull().astype(int)

print(df)
```
